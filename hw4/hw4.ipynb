```python
import math
import numpy as np
import cv2
from scipy import ndimage
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('sar_1.jpg')
image_gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY) 
plt.imshow(image_gray, cmap="gray")
```

```python
# Алгоритм разрастания

def homo_average(img, mask, point, T):
    av_val = img[mask > 0].sum() / np.count_nonzero(img[mask > 0])
                                                            
    if abs(av_val - img[point]) <= T:
        return True
    
    return False




def region_growing(image, seed_point,homo_fun,r, T):
    mask = np.zeros(image_gray.shape, np.uint8)
    mask[seed_point] = 1
    count = 1
    while count > 0:
        count = 0
        local_mask = np.zeros(image_gray.shape, np.uint8)
        for i in range(r,image.shape[0] - r):
            for j in range(r,image.shape[1] - r):
                if mask[i,j]==0 and mask[i - r:i + r, j-r: j+r].sum() > 0:
                    if homo_fun(image, mask, (i,j), T):
                        local_mask[i,j] = 1
        count = np.count_nonzero(local_mask)
        print(count)
        mask += local_mask
        
    return mask*255
```

```python
seed_point = (250,250)
mask = region_growing(image_gray,seed_point,homo_average, 7, 20)
plt.imshow(mask, cmap="gray")
```

```python
# вычисление отличного критерия однородности

def homo_local(img, mask, point, T):
    i, j = point
    r = 2
    
    neighborhood = img[max(0, i-r):i+r+1, max(0, j-r):j+r+1]
    av_val = neighborhood.mean()
    
    if abs(av_val - img[point]) <= T:
        return True
    
    return False
```

```python
seed_point = (250, 250)
mask_local = region_growing(image_gray, seed_point, homo_local, 7, 8)
plt.imshow(mask_local, cmap="gray")
```

```python
# алгоритм сегментации watershed+distance transform для задачи подсчета пальмовых деревьев.
image = cv2.imread('palm_1.jpg')
image_gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
plt.imshow(image_gray, cmap="gray")

```

```python
ret, thresh = cv2.threshold(image_gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
plt.imshow(thresh, cmap="gray")
```

```python
dist = cv2.distanceTransform(thresh, cv2.DIST_L2, 5) 
plt.imshow(dist, cmap="gray")
```

```python
ret, sure_fg = cv2.threshold(dist, 0.39 * dist.max(), 255, cv2.THRESH_BINARY)
plt.imshow(sure_fg, cmap="gray")

```

```python
sure_fg = sure_fg.astype(np.uint8)
ret, markers = cv2.connectedComponents(sure_fg) 
print("число пальм:", ret - 1)

```

```python
markers = cv2.watershed(image, markers)
plt.imshow(markers, cmap="gray")

print("Итог", markers.max())
```
